```python
import pandas as pd
```

```python
def get_texts(file_path: str, max_percent: int = 10) -> str:
    with open(file_path) as f:
        text = f.read()

    size = len(text)
    return text[: int(size * max_percent // 100)]
```

```python
EN_PATH = "/Users/saranyapal/Developer/Neural-Machine-Translation/data/wat-2021/test.en"
BN_PATH = "/Users/saranyapal/Developer/Neural-Machine-Translation/data/wat-2021/test.bn"
HI_PATH = "/Users/saranyapal/Developer/Neural-Machine-Translation/data/wat-2021/test.hi"
```

```python
en_texts = get_texts(EN_PATH, 100)
bn_texts = get_texts(BN_PATH, 100)
hi_texts = get_texts(HI_PATH, 100)
```

```python
en_texts_list = en_texts.split("\n")
bn_texts_list = bn_texts.split("\n")
hi_texts_list = hi_texts.split("\n")    
```
